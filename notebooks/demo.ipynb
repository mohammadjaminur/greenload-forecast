# GreenLoad reproducible experiment

This notebook reads the committed metrics and plots without requiring local checkpoints or numerical arrays.
To run an independent experiment, use `python -m greenload.experiments --budget-hours 4 --run-dir results/experiments/reproduction` from the repository root.
The published selection was frozen before final testing; this notebook does not tune against test results.

```python
from pathlib import Path
import json

root = Path("../results/experiments/tuning-180d")
selection = json.loads((root / "selection.json").read_text())
run = root / selection["run_id"]
print(json.dumps(selection, indent=2))
```

```python
metrics = json.loads((run / "test_gru_metrics.json").read_text())
print("MPC:", metrics["scheduling"])
for signal, score in metrics["forecast"].items():
    print(signal, "NRMSE:", score["nrmse"])
```

```python
import matplotlib.pyplot as plt

for name in ("example_test_forecast.png", "test_mpc_cost.png", "test_forecast_rmse.png"):
    fig, ax = plt.subplots(figsize=(12, 6))
    ax.imshow(plt.imread(root / "figures" / name))
    ax.axis("off")
    fig.tight_layout()
```
